```python
orders["order_date"] = pd.to_datetime(orders["order_date"],format="%Y-%m-%d")
```

```python
analysis_date =orders["order_date"].max() + pd.Timedelta(days=1)
analysis_date
```

```python
rfm=pd.merge(orders,order_items,on='order_id',how='right')
rfm
```

```python
analysis_datas=orders['order_date'].max()
analysis_datas
```

```python
frequency=rfm.groupby('customer_id')['order_id'].nunique()
frequency
```

```python
monetary=rfm.groupby('customer_id')['List_Price'].sum()
monetary
```

```python
last_date=rfm.groupby('customer_id')['order_date'].max()
last_date
```

```python
recency=(analysis_date-last_date).dt.days
recency
```

```python
rfm= rfm.groupby("customer_id").agg(
    recency=("order_date", lambda x: (analysis_datas -x.max()).days),
    frequency=("order_id", "nunique"),
    monetary=("Total_Price", "sum")

).reset_index()
```

```python
rfm
```

```python
rfm['R_Score']=pd.qcut(rfm['recency'],4,labels=[4,3,2,1])
rfm['F_Score']=pd.qcut(rfm['frequency'].rank(method='first'),4,labels=[1,2,3,4])
rfm['M_Score']=pd.qcut(rfm['monetary'].rank(method='first'),4,labels=[1,2,3,4])
```

```python
rfm['rfm_score']=(rfm['R_Score'].astype(int)+
                  rfm['F_Score'].astype(int)+
                  rfm['M_Score'].astype(int))
```

```python
rfm
```

```python
def customer_segment(score):
    if score >= 10:
        return 'Champions'
    elif score >= 8:
        return 'Loyal customer'
    elif score >= 6:
        return 'Potential loyalists'
    elif score >= 4:
        return 'At risk'
    else:
        return 'Lost customer'
    rfm['Segment']=rfm['rfm_score'].apply(customer_segment)
```

```python
rfm['Segment']=rfm['rfm_score'].apply(customer_segment)
```

```python
rfm
```

```python
rfm.to_sql(
    'customer_segment',
    con=caps,
    if_exists='replace',
    index=False)
```
